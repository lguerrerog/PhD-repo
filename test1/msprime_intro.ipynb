```python
import msprime
import drawsvg as draw
from IPython.display import display, SVG

# 1. Simular el árbol
ts = msprime.sim_ancestry(
    2,
    population_size=10_000,
    random_seed=42
)

# 2. Generar el SVG desde msprime
svg = ts.draw_svg(
    y_axis=True,
    size=(600, 300)
)

# 3. Guardar el SVG
with open("arbol_msprime.svg", "w", encoding="utf-8") as f:
    f.write(svg)

# 4. Mostrarlo en Jupyter
display(SVG(filename="arbol_msprime.svg"))
```

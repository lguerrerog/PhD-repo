# Basic Model for Archaic Population Substructure in Native Americans
In this first, basic model, the backbone of the Out of Africa > Beringia > America mirgation will be simulated, leaving complex demographic events aside for now.

Populations in this simulation:
* ANC: ancestral human population
* AFR: population that stays in Africa, child of ANC
* OOA: population that left Africa, child of ANC
* EAS: ancestors of East Asians, children of OOA
* BER: ancestors that will become Native Americans via Bringia, children of OOA
* NAT: the founding Native American population, children of BER

This simulation is not modeling migration between populations, multiple waves, introgressions, recent admixture. This will be added later.

```python
# Libraries that will be needed 
import msprime
import json
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Defining units of generations as 29 years per generation.
GEN_TIME = 29   # years per generation

# Converting from years to generations
def years_to_gens(years):
    return int(years / GEN_TIME)


# Setting up the times of population splits
T_OOA = years_to_gens(70_000)    # ANC -> AFR / OOA split
T_EAS_BER = years_to_gens(40_000)   # OOA -> EAS / BER split
T_NAT_FOUND = years_to_gens(16_000)   # BER -> NAT split
T_BOTTLENECK_START = years_to_gens(25_000)   # start of Beringian standstill bottleneck

# ----------------------------------------------------------------
# DEMOGRAPHY OBJECT
# ----------------------------------------------------------------

# Adding all the populations that existed
demography  = msprime.Demography()
demography.add_population(name="ANC", initial_size=10_000, description="Deep ancestral population")
demography.add_population(name="AFR", initial_size=40_000, description="Population remaining in Africa")
demography.add_population(name="OOA", initial_size=2_000, description="Out of Africa population, bottlenecked")
demography.add_population(name="EAS", initial_size=10_000, description="Ancestors of East Asians")
demography.add_population(name="BER", initial_size=2_000, description="Ancestral Beringian population")
demography.add_population(name="NAT", initial_size=5_000, description="Founding Native American population")

#---------------------------------------------------------------
# Demographic Events
# --------------------------------------------------------------
# Add demographic events as splits and bottlenecks
#In msprime events are defined going backwards in time

# Native Americans are derived from Beringian
demography.add_population_split(time=T_NAT_FOUND, derived=["NAT"], ancestral = "BER")

# Bottleneck of the Beringian standstill defined as a change of Ne. We define the size before the bottleneck
demography.add_population_parameters_change(time=T_BOTTLENECK_START, population="BER", initial_size=800)

# EAS and BER descend from OOA populations
demography.add_population_split(time=T_EAS_BER, derived=["EAS", "BER"], ancestral="OOA")

# OOA and AFR descend from the deep ancestral population ANC
demography.add_population_split(time=T_OOA, derived=["OOA", "AFR"], ancestral="ANC")

# Define migration
MIGRATION_RATE = 1e-5    # Rate per generation for low-level migration (1e-5 to 1e-3)

demography.set_symmetric_migration_rate(["AFR", "OOA"], rate=MIGRATION_RATE)
demography.set_symmetric_migration_rate(["EAS", "BER"], rate=MIGRATION_RATE)

# To sort all events according to time automatically (sanity check)
demography.sort_events()

# Print a table with the model specifications (sanity check, debug)
print(demography.debug())

# ----------------------------------------------------------------
# Simulation with migration in demography
# ----------------------------------------------------------------
ts = msprime.sim_ancestry(samples={"NAT": 10, "EAS":10},
                         demography=demography,
                         sequence_length=1_000_000,   # Toy genomic region, not full
                         recombination_rate=1e-8,
                         random_seed=44)    # Human recombination rate

# mutations
mts = msprime.sim_mutations(ts, rate=1.25e-8, random_seed=42)
```

```python
# Statistics for comparison
nat_pop_id = [p.id for p in demography.populations if p.name == "NAT"][0]
eas_pop_id = [p.id for p in demography.populations if p.name == "EAS"][0]
 
nat_samples = mts.samples(population=nat_pop_id)
eas_samples = mts.samples(population=eas_pop_id)
 
pi_nat = mts.diversity(sample_sets=nat_samples)
pi_eas = mts.diversity(sample_sets=eas_samples)
dxy_nat_eas = mts.divergence(sample_sets=[nat_samples, eas_samples])
fst_nat_eas = mts.Fst(sample_sets=[nat_samples, eas_samples])
tajd_nat = mts.Tajimas_D(sample_sets=nat_samples)
tajd_eas = mts.Tajimas_D(sample_sets=eas_samples)
sfs_nat = mts.allele_frequency_spectrum(sample_sets=[nat_samples], polarised=False, span_normalise=False)
sfs_eas = mts.allele_frequency_spectrum(sample_sets=[eas_samples], polarised=False, span_normalise=False)
 
results = {
    "model_version": "v2_with_migration",
    "description": (
        "Same as v1 (OOA -> EAS/BER split -> Beringian bottleneck -> NAT "
        f"founding), plus symmetric migration (rate={MIGRATION_RATE}) "
        "between AFR<->OOA and EAS<->BER while each pair coexists"
    ),
    "diversity_NAT": pi_nat,
    "diversity_EAS": pi_eas,
    "divergence_NAT_EAS": dxy_nat_eas,
    "Fst_NAT_EAS": fst_nat_eas,
    "TajimasD_NAT": tajd_nat,
    "TajimasD_EAS": tajd_eas,
    "SFS_NAT": sfs_nat.tolist(),
    "SFS_EAS": sfs_eas.tolist(),
}
 
print("\n=== Summary statistics: v2 (with migration) ===")
print("-- Nucleotide diversity (pi) --")
print(f"Nucleotide diversity (pi)   NAT: {pi_nat:.6f}")
print(f"Nucleotide diversity (pi)   EAS: {pi_eas:.6f}")
print("-- Divergence (Dxy) --")
print(f"Divergence (Dxy) NAT vs EAS:     {dxy_nat_eas:.6f}")
print("-- Fst --")
print(f"Fst NAT vs EAS:                  {fst_nat_eas:.4f}")
print("-- Tajima's D --")
print(f"Tajima's D          NAT:         {tajd_nat:.4f}")
print(f"Tajima's D          EAS:         {tajd_eas:.4f}")
print(f"SFS NAT (folded, raw counts):    {sfs_nat}")
print(f"SFS EAS (folded, raw counts):    {sfs_eas}")


# Append to json file
with open("model_comparison_stats.jsonl", "a") as f:
    f.write(json.dumps(results) + "\n")
 
print("\nAppended results to model_comparison_stats.jsonl")
```

```python
# Plotting SFS

x = np.arange(len(sfs_nat))
width = 0.4

plt.figure(figsize=(12, 5))

plt.bar(x - width/2, sfs_nat, width, label='SFS NAT')
plt.bar(x + width/2, sfs_eas, width, label='SFS EAS')

plt.xlabel('Frequency')
plt.ylabel('Number of variants')
plt.xticks(x)
plt.xlim(right=11)
plt.legend()
plt.tight_layout()
plt.savefig("SFS_nat_eas_v2_10")

plt.show()
```

# Model v3 - Migration and Africa included in sampling
Changes in this model: 
* include AFR in the samples
* Each sample has 50 diploid individuals
* Initial population sizes changed for AFR and EAS to 40000 and 10000 respectivelly.
* fixed bug with migration in populations not existing at time 0 using add_migration_rate_change() instead of set_symmetric_migration_rate()

```python
# Libraries that will be needed 
import msprime
import json
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Defining units of generations as 29 years per generation.
GEN_TIME = 29   # years per generation

# Converting from years to generations
def years_to_gens(years):
    return int(years / GEN_TIME)

# Setting up the times of population splits
T_OOA = years_to_gens(70_000)    # ANC -> AFR / OOA split
T_EAS_BER = years_to_gens(40_000)   # OOA -> EAS / BER split
T_NAT_FOUND = years_to_gens(16_000)   # BER -> NAT split
T_BOTTLENECK_START = years_to_gens(25_000)   # start of Beringian standstill bottleneck

# ----------------------------------------------------------------
# DEMOGRAPHY OBJECT
# ----------------------------------------------------------------

# Adding all the populations that existed
demography  = msprime.Demography()
demography.add_population(name="ANC", initial_size=10_000, description="Deep ancestral population")
demography.add_population(name="AFR", initial_size=40_000, description="Population remaining in Africa")
demography.add_population(name="OOA", initial_size=2_000, description="Out of Africa population, bottlenecked")
demography.add_population(name="EAS", initial_size=10_000, description="Ancestors of East Asians")
demography.add_population(name="BER", initial_size=2_000, description="Ancestral Beringian population")
demography.add_population(name="NAT", initial_size=5_000, description="Founding Native American population")

#---------------------------------------------------------------
# Demographic Events
# --------------------------------------------------------------
# Add demographic events as splits and bottlenecks
#In msprime events are defined going backwards in time

# Native Americans are derived from Beringian
demography.add_population_split(time=T_NAT_FOUND, derived=["NAT"], ancestral = "BER")

# Bottleneck of the Beringian standstill defined as a change of Ne. We define the size before the bottleneck
demography.add_population_parameters_change(time=T_BOTTLENECK_START, population="BER", initial_size=800)

# EAS and BER descend from OOA populations
demography.add_population_split(time=T_EAS_BER, derived=["EAS", "BER"], ancestral="OOA")

# OOA and AFR descend from the deep ancestral population ANC
demography.add_population_split(time=T_OOA, derived=["OOA", "AFR"], ancestral="ANC")

# Define migration. Uncomment this three lines to include migration in the model
MIGRATION_RATE = 1e-5    # Rate per generation for low-level migration (1e-5 to 1e-3)

# EAS <-> BER migration at T_NAT_FOUND time. Symmetric migration
demography.add_migration_rate_change(time=T_NAT_FOUND, source="EAS", dest="BER", rate=MIGRATION_RATE)
demography.add_migration_rate_change(time=T_NAT_FOUND, source="BER", dest="EAS", rate=MIGRATION_RATE)

# OOA <-> AFR migration at T_EAS_BER time. Symmetric migration.
demography.add_migration_rate_change(time=T_EAS_BER, source="OOA", dest="AFR", rate=MIGRATION_RATE)
demography.add_migration_rate_change(time=T_EAS_BER, source="AFR", dest="OOA", rate=MIGRATION_RATE)

# To sort all events according to time automatically (sanity check)
demography.sort_events()

# Print a table with the model specifications (sanity check, debug)
print(demography.debug())

#---------------------------------------------------------------
# Simulation with migration in demography
#---------------------------------------------------------------
ts = msprime.sim_ancestry(samples={"NAT": 50, "EAS":50, "AFR":50},
                         demography=demography,
                         sequence_length=1_000_000,   # Toy genomic region, not full
                         recombination_rate=1e-8, # Human recombination rate
                         random_seed=44)    

# mutations
mts = msprime.sim_mutations(ts, rate=1.25e-8, random_seed=42)
```

```python
# Statistics for comparison

nat_pop_id = [p.id for p in demography.populations if p.name == "NAT"][0]
eas_pop_id = [p.id for p in demography.populations if p.name == "EAS"][0]
afr_pop_id = [p.id for p in demography.populations if p.name == "AFR"][0]
 
nat_samples = mts.samples(population=nat_pop_id)
eas_samples = mts.samples(population=eas_pop_id)
afr_samples = mts.samples(population=afr_pop_id)

# Diversity
pi_nat = mts.diversity(sample_sets=nat_samples)
pi_eas = mts.diversity(sample_sets=eas_samples)
pi_afr = mts.diversity(sample_sets=afr_samples)

# Divergence
dxy_nat_eas = mts.divergence(sample_sets=[nat_samples, eas_samples])
dxy_nat_afr = mts.divergence(sample_sets=[nat_samples, afr_samples])
dxy_afr_eas = mts.divergence(sample_sets=[afr_samples, eas_samples])

fst_nat_eas = mts.Fst(sample_sets=[nat_samples, eas_samples])
fst_nat_afr = mts.Fst(sample_sets=[nat_samples, afr_samples])
fst_afr_eas = mts.Fst(sample_sets=[afr_samples, eas_samples])

# Tajima's D
tajd_nat = mts.Tajimas_D(sample_sets=nat_samples)
tajd_eas = mts.Tajimas_D(sample_sets=eas_samples)
tajd_afr = mts.Tajimas_D(sample_sets=afr_samples)
# SFS
sfs_nat = mts.allele_frequency_spectrum(sample_sets=[nat_samples], polarised=False, span_normalise=False)
sfs_eas = mts.allele_frequency_spectrum(sample_sets=[eas_samples], polarised=False, span_normalise=False)
sfs_afr = mts.allele_frequency_spectrum(sample_sets=[afr_samples], polarised=False, span_normalise=False)
 
results = {
    "model_version": "v3_afr_samples_larger_n",
    "description": (
        "Same topology as v2, plus: AFR sampled directly, sample sizes "
        "50/pop, AFR Ne=40000, EAS Ne=10000, migration rates fixed to "
        "start at each population's activation time instead of time 0"
    ),
    "diversity_NAT": pi_nat,
    "diversity_EAS": pi_eas,
    "diversity_AFR": pi_afr,
    "divergence_NAT_EAS": dxy_nat_eas,
    "divergence_NAT_AFR": dxy_nat_afr,
    "divergence_AFR_EAS": dxy_afr_eas,
    "Fst_NAT_EAS": fst_nat_eas,
    "Fst_NAT_AFR": fst_nat_afr,
    "Fst_AFR_EAS": fst_afr_eas,
    "TajimasD_NAT": tajd_nat,
    "TajimasD_EAS": tajd_eas,
    "TajimasD_AFR": tajd_afr,
    "SFS_NAT": sfs_nat.tolist(),
    "SFS_EAS": sfs_eas.tolist(),
    "SFS_AFR": sfs_afr.tolist(),
}

 
print("\n=== Summary statistics: v3 (with migration) ===")
print("-- Nucleotide diversity (pi) --")
print(f"Nucleotide diversity (pi)   NAT: {pi_nat:.6f}")
print(f"Nucleotide diversity (pi)   EAS: {pi_eas:.6f}")
print(f"Nucleotide diversity (pi)   AFR: {pi_afr:.6f}")
print("-- Divergence (Dxy) --")
print(f"Divergence (Dxy) NAT vs EAS:     {dxy_nat_eas:.6f}")
print(f"Divergence (Dxy) NAT vs AFR:     {dxy_nat_afr:.6f}")
print(f"Divergence (Dxy) AFR vs EAS:     {dxy_afr_eas:.6f}")
print("-- Fst --")
print(f"Fst NAT vs EAS:                  {fst_nat_eas:.4f}")
print(f"Fst NAT vs AFR:                  {fst_nat_afr:.4f}")
print(f"Fst AFR vs EAS:                  {fst_afr_eas:.4f}")
print("-- Tajima's D --")
print(f"Tajima's D          NAT:         {tajd_nat:.4f}")
print(f"Tajima's D          EAS:         {tajd_eas:.4f}")
print(f"Tajima's D          AFR:         {tajd_afr:.4f}")
print("-- SFS raw counts --")
print(f"SFS NAT (folded, raw counts):    {sfs_nat}")
print(f"SFS EAS (folded, raw counts):    {sfs_eas}")
print(f"SFS AFR (folded, raw counts):    {sfs_afr}")


# Append to json file
with open("model_comparison_stats.jsonl", "a") as f:
    f.write(json.dumps(results) + "\n")
 
print("\nAppended results to model_comparison_stats.jsonl")
```

```python
fig, axes = plt.subplots(
    1, 3,
    figsize=(15, 6),
    sharex=False,
    sharey=True
)

sfs_samples = [sfs_afr, sfs_eas, sfs_nat]
names = ["AFR", "EAS", "NAT"]

axes = axes.ravel()

for ax, sss, n in zip(axes, sfs_samples, names):
    values = sss

    # Find the last non-zero frequency and trim everything after it
    nonzero = np.flatnonzero(values != 0)

    if len(nonzero) > 0:
        values = values[:nonzero[-1] + 1]

    x = np.arange(len(values))

    ax.bar(
        x,
        values,
        width=0.9,
        edgecolor="black",
        linewidth=0.5
    )

    ax.set_title(n, fontsize=14, fontweight="bold")
    ax.set_xlabel("Allele frequency", fontsize=11)
    ax.set_ylabel("SFS count", fontsize=11)

    ax.set_xlim(-0.5, len(values) - 0.5)
    ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig("SFS_v3.png", dpi=300, bbox_inches='tight')
plt.show()
```

# Model v4 - Adding split between North and South America 

```python
# Libraries that will be needed 
import msprime
import json
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Defining units of generations as 29 years per generation.
GEN_TIME = 29   # years per generation

# Converting from years to generations
def years_to_gens(years):
    return int(years / GEN_TIME)


# Setting up the times of population splits
T_OOA = years_to_gens(70_000)    # ANC -> AFR / OOA split
T_EAS_BER = years_to_gens(40_000)   # OOA -> EAS / BER split
T_NAT_FOUND = years_to_gens(16_000)   # BER -> NAT split
T_BOTTLENECK_START = years_to_gens(25_000)   # start of Beringian standstill bottleneck
T_NNA_SAM = years_to_gens(14_000)   # As an average of papers from Moreno-Mayas, 2018 and Rider, 2024

# ----------------------------------------------------------------
# DEMOGRAPHY OBJECT
# ----------------------------------------------------------------

# Adding all the populations that existed
demography  = msprime.Demography()
demography.add_population(name="ANC", initial_size=10_000, description="Deep ancestral population")
demography.add_population(name="AFR", initial_size=40_000, description="Population remaining in Africa")
demography.add_population(name="OOA", initial_size=2_000, description="Out of Africa population, bottlenecked")
demography.add_population(name="EAS", initial_size=10_000, description="Ancestors of East Asians")
demography.add_population(name="BER", initial_size=2_000, description="Ancestral Beringian population")
demography.add_population(name="NAT", initial_size=5_000, description="Founding Native American population")
demography.add_population(name="NNA", initial_size=6_000, description="Natvie North America population")
demography.add_population(name="SAM", initial_size=600, description="South America population")

#---------------------------------------------------------------
# Demographic Events
# --------------------------------------------------------------
# Add demographic events as splits and bottlenecks
#In msprime events are defined going backwards in time

demography.add_population_split(time=T_NNA_SAM, derived=["NNA", "SAM"], ancestral="NAT")

# Native Americans are derived from Beringian
demography.add_population_split(time=T_NAT_FOUND, derived=["NAT"], ancestral="BER")

# Bottleneck of the Beringian standstill defined as a change of Ne. We define the size before the bottleneck
demography.add_population_parameters_change(time=T_BOTTLENECK_START, population="BER", initial_size=800)

# EAS and BER descend from OOA populations
demography.add_population_split(time=T_EAS_BER, derived=["EAS", "BER"], ancestral="OOA")

# OOA and AFR descend from the deep ancestral population ANC
demography.add_population_split(time=T_OOA, derived=["OOA", "AFR"], ancestral="ANC")

# Define migration. Uncomment this three lines to include migration in the model
MIGRATION_RATE = 1e-5    # Rate per generation for low-level migration (1e-5 to 1e-3)

# EAS <-> BER migration at T_NAT_FOUND time. Symmetric migration
demography.add_migration_rate_change(time=T_NAT_FOUND, source="EAS", dest="BER", rate=MIGRATION_RATE)
demography.add_migration_rate_change(time=T_NAT_FOUND, source="BER", dest="EAS", rate=MIGRATION_RATE)

# OOA <-> AFR migration at T_EAS_BER time. Symmetric migration.
demography.add_migration_rate_change(time=T_EAS_BER, source="OOA", dest="AFR", rate=MIGRATION_RATE)
demography.add_migration_rate_change(time=T_EAS_BER, source="AFR", dest="OOA", rate=MIGRATION_RATE)

# To sort all events according to time automatically (sanity check)
demography.sort_events()

# Print a table with the model specifications (sanity check, debug)
print(demography.debug())

###---------------------------------------------------------
# Run simulation
ts = msprime.sim_ancestry(
    samples={"NNA": 50, "SAM": 50, "EAS": 50, "AFR": 50},
    demography=demography, sequence_length=1_000_000,
    recombination_rate=1e-8, random_seed=44
)
print("simulation ran. Number of trees=", ts.num_trees)

# Add mutations
mts = msprime.sim_mutations(ts, rate=1.25e-8, random_seed=42)
```

```python
# ---------Statistics--------------
# Get all population ids
pop_ids = {name: [p.id for p in demography.populations if p.name == name][0]
           for name in ["AFR", "EAS", "NNA", "SAM"]}
# Get the samples
samples = {name: mts.samples(population=pid) for name, pid in pop_ids.items()}

# Calculate diversity and divergence
diversity = {name: mts.diversity(sample_sets=s) for name, s in samples.items()}
tajimas_d = {name: mts.Tajimas_D(sample_sets=s) for name, s in samples.items()}

# Calculate Dxy and Fst
pairs = [("NNA", "SAM"), ("AFR", "NNA"), ("AFR", "SAM"), ("EAS", "NNA"), ("EAS", "SAM"), ("AFR", "EAS")]
dxy = {f"{a}-{b}": mts.divergence(sample_sets=[samples[a], samples[b]]) for a, b in pairs}
fst = {f"{a}-{b}": mts.Fst(sample_sets=[samples[a], samples[b]]) for a, b in pairs}

# Calculate SFS
sfs = {name: mts.allele_frequency_spectrum(sample_sets=[f], polarised=False, span_normalise=False) for name, f in samples.items()}

# Results
print("\n=== Summary statistics: v4 (NAM/SAM split) ===")
print("-- Nucleotide diversity (pi) --")
for k, v in diversity.items():
    print(f"  {k}: {v:.6f}")
print("-- Tajima's D --")
for k, v in tajimas_d.items():
    print(f"  {k}: {v:.4f}")
print("-- Divergence (Dxy) --")
for k, v in dxy.items():
    print(f"  {k}: {v:.6f}")
print("-- Fst --")
for k, v in fst.items():
    print(f"  {k}: {v:.4f}")
print("-- SFS --")
for k, v in sfs.items():
    print(f"  {k}: {v}")
```

```python
fig, axes = plt.subplots(
    2, 2,
    figsize=(14, 9),
    sharex=False,
    sharey=True
)

axes = axes.ravel()

for ax, (name, values) in zip(axes, sfs.items()):
    values = np.asarray(values)

    # Find the last non-zero frequency and trim everything after it
    nonzero = np.flatnonzero(values != 0)

    if len(nonzero) > 0:
        values = values[:nonzero[-1] + 1]

    x = np.arange(len(values))

    ax.bar(
        x,
        values,
        width=0.9,
        edgecolor="black",
        linewidth=0.5
    )

    ax.set_title(name, fontsize=14, fontweight="bold")
    ax.set_xlabel("Allele frequency", fontsize=11)
    ax.set_ylabel("SFS count", fontsize=11)

    ax.set_xlim(-0.5, len(values) - 0.5)
    ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig("SFS_v4.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, axes = plt.subplots(
    3, 1,
    figsize=(6, 9)
)
# Diversity
names = list(diversity.keys())
values = list(diversity.values())

axes[0].bar(
    names,
    values,
    edgecolor="black",
    linewidth=0.8
)

axes[0].set_title(
    "Nucleotide diversity (π)",
    fontsize=14,
    fontweight="bold"
)

axes[0].set_ylabel("Diversity (π)")
axes[0].grid(axis="y", alpha=0.3)

# Tajima's D
names = list(tajimas_d.keys())
values = list(tajimas_d.values())

axes[1].bar(
    names,
    values,
    edgecolor="black",
    linewidth=0.8
)

axes[1].set_title(
    "Tajima's D",
    fontsize=14,
    fontweight="bold"
)

axes[1].set_ylabel("Tajima's D")
axes[1].axhline(
    0,
    linewidth=1,
    linestyle="--"
)

axes[1].grid(axis="y", alpha=0.3)

# Divergence
names = list(dxy.keys())
values = list(dxy.values())

axes[2].bar(
    names,
    values,
    edgecolor="black",
    linewidth=0.8
)

axes[2].set_title(
    "Between-population divergence (dxy)",
    fontsize=14,
    fontweight="bold"
)

axes[2].set_ylabel("Divergence (dxy)")
axes[2].grid(axis="y", alpha=0.3)

for ax in axes:
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.savefig("div_v4.png", dpi=300, bbox_inches='tight')
plt.show()
```

# Model v5 - corrections
* Include migration between NNA and SAM
* Modification of NAT population size from 5000 to 1791 before NNA - SAM split. Taken from Rider et al., 2024
* Ran several replicates to get distribution of the statistics. Runs are independent (different random seeds). Number of runs: 20.
* Statistics reported as the mean +/- standard deviation. This way, results are possible to distinguish from noise. 

```python
import json
import statistics
import msprime
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define demography

GEN_TIME = 29
 
def years_to_gens(years):
    return int(years / GEN_TIME)
 
T_OOA = years_to_gens(70_000)
T_EAS_BER = years_to_gens(40_000)
T_BOTTLENECK_START = years_to_gens(25_000)
T_NAT_FOUND = years_to_gens(16_000)
T_NNA_SAM = years_to_gens(14_000)
 
MIGRATION_RATE = 1e-5

# Function to define a fresh demography object for every run. 
# Needed because msprime objects accumulate state so calls would not be independent.
def build_demography():
    demography = msprime.Demography()
    demography.add_population(name="ANC", initial_size=10_000)
    demography.add_population(name="AFR", initial_size=40_000)
    demography.add_population(name="OOA", initial_size=2_000)
    demography.add_population(name="EAS", initial_size=10_000)
    demography.add_population(name="BER", initial_size=2_000)
    demography.add_population(name="NAT", initial_size=1791)   # Changed the size of the population size before the American split
    demography.add_population(name="NNA", initial_size=6000)
    demography.add_population(name="SAM", initial_size=600)

    # Add demographic events
    demography.add_population_split(time=T_NNA_SAM, derived=["NNA", "SAM"], ancestral="NAT")
    demography.add_population_split(time=T_NAT_FOUND, derived=["NAT"], ancestral="BER")
    demography.add_population_parameters_change(time=T_BOTTLENECK_START, population="BER", initial_size=800)
    demography.add_population_split(time=T_EAS_BER, derived=["EAS", "BER"], ancestral="OOA")
    demography.add_population_split(time=T_OOA, derived=["OOA", "AFR"], ancestral="ANC")

    # Add migrations
    demography.set_symmetric_migration_rate(["NNA", "SAM"], rate=MIGRATION_RATE)
    demography.add_migration_rate_change(time=T_NAT_FOUND, source="BER", dest="EAS", rate=MIGRATION_RATE)
    demography.add_migration_rate_change(time=T_NAT_FOUND, source="EAS", dest="BER", rate=MIGRATION_RATE)
    demography.add_migration_rate_change(time=T_EAS_BER, source="OOA", dest="AFR", rate=MIGRATION_RATE)
    demography.add_migration_rate_change(time=T_EAS_BER, source="AFR", dest="OOA", rate=MIGRATION_RATE)

    # Sanity check
    demography.sort_events()
    return demography

# Check the debug
print(build_demography().debug())

# Run N=20 replicates and collect statistics
N = 20

stats_names = [
    "diversity_AFR", "diversity_EAS", "diversity_NNA", "diversity_SAM",
    "TajimasD_AFR", "TajimasD_EAS", "TajimasD_NNA", "TajimasD_SAM",
    "Dxy_NNA-SAM", "Dxy_EAS-NNA", "Dxy_EAS-SAM", "Dxy_AFR-NNA", "Dxy_AFR-SAM", "Dxy_AFR-EAS",
    "Fst_NNA-SAM", "Fst_EAS-NNA", "Fst_EAS-SAM", "Fst_AFR-NNA", "Fst_AFR-SAM", "Fst_AFR-EAS",
    
]

# Collect values for each replication
replicate_values = {name:[] for name in stats_names}

for seed in range(1, N+1):
    demography = build_demography()
    ts = msprime.sim_ancestry(
        samples={"NNA": 50, "SAM": 50, "EAS": 50, "AFR": 50},
        demography=demography,
        sequence_length=1_000_000,
        random_seed=seed
    )
    #Mutation seed also changing. Use different seed so ts and mts don't share a stream accidentally 
    mts = msprime.sim_mutations(ts, rate=1.25e-8, random_seed=seed + 10_000)
    
    pop_ids = {name: [p.id for p in demography.populations if p.name == name][0]
               for name in ["AFR", "EAS", "NNA", "SAM"]}

    samples = {name: mts.samples(population=pid) for name, pid in pop_ids.items()}

    for name, s in samples.items():
        # Diversity (pi)
        replicate_values[f"diversity_{name}"].append(mts.diversity(sample_sets=s))
        # Tajima's D
        replicate_values[f"TajimasD_{name}"].append(mts.Tajimas_D(sample_sets=s))
        # SFS
        #replicate_values[f"sfs_{name}"].append(mts.allele_frequency_spectrum(sample_sets=[s], polarised=False, span_normalise=False))
        
    # Divergence and Fst
    pairs = [("NNA", "SAM"), ("EAS", "NNA"), ("EAS", "SAM"), ("AFR", "NNA"), ("AFR", "SAM"), ("AFR", "EAS")]
    for a, b in pairs:
        replicate_values[f"Dxy_{a}-{b}"].append(mts.divergence(sample_sets=[samples[a], samples[b]]))
        replicate_values[f"Fst_{a}-{b}"].append(mts.Fst(sample_sets=[samples[a], samples[b]]))


# Mean and Standard deviation across replicates

summary = {}
for name, values in replicate_values.items():
    summary[f"{name}_mean"] = statistics.mean(values)
    summary[f"{name}_std"] = statistics.stdev(values)

print(f"\n=== Summary statistics: v5, mean +/- std across {N} replicates ===")
print("-- Nucleotide diversity (pi) --")
for name in ["AFR", "EAS", "NNA", "SAM"]:
    m, s = summary[f"diversity_{name}_mean"], summary[f"diversity_{name}_std"]
    print(f"{name}:{m:.6f} +/- {s:.6f}")

print("-- Tajima's D --")
for name in ["AFR", "EAS", "NNA", "SAM"]:
    m, s = summary[f"TajimasD_{name}_mean"], summary[f"TajimasD_{name}_std"]
    print(f"  {name}: {m:.4f} +/- {s:.4f}")
    
print("-- Divergence (Dxy) --")
for a, b in [("NNA", "SAM"), ("EAS", "NNA"), ("EAS", "SAM"), ("AFR", "NNA"), ("AFR", "SAM"), ("AFR", "EAS")]:
    m, s = summary[f"Dxy_{a}-{b}_mean"], summary[f"Dxy_{a}-{b}_std"]
    print(f"  {a}-{b}: {m:.6f} +/- {s:.6f}")
    
print("-- Fst --")
for a, b in [("NNA", "SAM"), ("EAS", "NNA"), ("EAS", "SAM"), ("AFR", "NNA"), ("AFR", "SAM"), ("AFR", "EAS")]:
    m, s = summary[f"Fst_{a}-{b}_mean"], summary[f"Fst_{a}-{b}_std"]
    print(f"  {a}-{b}: {m:.4f} +/- {s:.4f}")
 

# Save the results
results = {
    "model_version": "v5_replicated_nat_fix_nam_sam_migration",
    "description": (
        f"Same topology as v4, plus: NAT Ne corrected 5000->1791 "
        f"(Rider et al. 2024), NNA<->SAM migration added (rate="
        f"{MIGRATION_RATE}), and results now averaged over "
        f"{N} replicates (mean +/- std reported)"
    ),
    "n_replicates": N,
    **summary,
}
with open("model_comparison_stats.jsonl", "a") as f:
    f.write(json.dumps(results) + "\n")
 
# Raw per-replicate values, saved separately so the full distribution
# (not just mean/std) is available for future plotting if needed.
with open("v5_replicate_raw_values.json", "w") as f:
    json.dump(replicate_values, f, indent=2)
 
print("\nAppended summary to model_comparison_stats.jsonl")
print("Saved raw per-replicate values to v5_replicate_raw_values.json")
```

```python
"""
================================================================================
 PLOTTING SCRIPT - diversity and Tajima's D across the 20 v5 replicates
================================================================================
Project: Archaic population substructures in Native Americans (aDNA project)

WHAT THIS DOES
This script reads the raw, per-replicate values saved by
native_american_demography_v5.py (in v5_replicate_raw_values.json - the
file that keeps every individual replicate's value, not just the
mean +/- std summary that went into model_comparison_stats.jsonl) and
draws two scatter plots:
  1. Nucleotide diversity (pi) for all four populations (AFR, EAS, NAM, SAM)
     across the 20 replicates.
  2. Tajima's D for the same four populations across the same 20 replicates.

Each population gets its own color AND its own marker shape (circle,
square, triangle, diamond). Using two visual channels together, rather
than color alone, keeps the four populations distinguishable even for
colorblind readers or in grayscale print - important for a PhD figure
that may end up in a thesis or paper.


================================================================================
"""

import json

import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------------------
# 1. LOAD THE RAW PER-REPLICATE VALUES
# ------------------------------------------------------------------------
# Adjust this path if you've moved the file - it's the JSON produced
# alongside model_comparison_stats.jsonl when native_american_demography_v5.py
# was run, and it holds all 20 individual replicate values per statistic
# (not just their mean/std, which is all that went into the comparison log).
RAW_VALUES_PATH = "v5_replicate_raw_values.json"

with open(RAW_VALUES_PATH) as f:
    raw_values = json.load(f)

# ------------------------------------------------------------------------
# 2. POPULATIONS, COLORS, AND MARKER SHAPES
# ------------------------------------------------------------------------
POPULATIONS = ["AFR", "EAS", "NNA", "SAM"]

# One hex color and one marker shape per population - both fixed here so
# the same population always gets the same look across every figure you
# make from this script (never reassign colors dynamically based on
# whichever populations happen to be plotted - keeping the mapping fixed
# is what lets a reader learn "blue circle = AFR" once and reuse it).
COLORS = {
    "AFR": "#2a78d6",  # blue
    "EAS": "#eb6834",  # orange
    "NNA": "#1baf7a",  # aqua/green
    "SAM": "#4a3aa7",  # violet
}
MARKERS = {
    "AFR": "o",  # circle
    "EAS": "s",  # square
    "NNA": "^",  # triangle
    "SAM": "D",  # diamond
}

# ------------------------------------------------------------------------
# 3. REUSABLE PLOTTING FUNCTION
# ------------------------------------------------------------------------
def plot_statistic_across_replicates(key_prefix, ylabel, title, out_path):
    """
    Draws one scatter+trend-line figure for a given statistic.

    key_prefix : the prefix used in v5_replicate_raw_values.json for this
                 statistic's per-population keys, e.g. "diversity" reads
                 keys "diversity_AFR", "diversity_EAS", etc. (see how
                 native_american_demography_v5.py names them when saving).
    ylabel     : label for the y-axis (e.g. "Nucleotide diversity (pi)").
    title      : figure title.
    out_path   : where to save the resulting PNG.
    """
    fig, ax = plt.subplots(figsize=(8, 5), dpi=150)

    # Light, neutral background rather than pure white - easier on the
    # eyes and consistent with common scientific-figure styling.
    fig.patch.set_facecolor("#fcfcfb")
    ax.set_facecolor("#fcfcfb")

    for pop in POPULATIONS:
        values = np.array(raw_values[f"{key_prefix}_{pop}"])
        n_replicates = len(values)
        x = np.arange(1, n_replicates + 1)  # replicate numbers: 1, 2, ..., 20

        # --- the scatter points for this population ---
        ax.scatter(
            x, values,
            color=COLORS[pop], marker=MARKERS[pop],
            s=45, label=pop, zorder=3,
            edgecolors="white", linewidths=0.6,  # thin white ring helps
                                                   # overlapping points stay
                                                   # readable
        )

        # --- the trend line: simple linear least-squares fit ---
        # np.polyfit(x, y, 1) returns [slope, intercept] for the best-fit
        # straight line through the 20 points. We only need the two
        # endpoints to draw a straight line, so we evaluate the fit at
        # the smallest and largest x value.
        slope, intercept = np.polyfit(x, values, 1)
        x_trend = np.array([x.min(), x.max()])
        y_trend = slope * x_trend + intercept
        ax.plot(
            x_trend, y_trend,
            color=COLORS[pop], linewidth=2, alpha=0.85, zorder=2,
        )

    # --- axis labels, title, ticks ---
    ax.set_xlabel("Replicate #")
    ax.set_ylabel(ylabel)
    ax.set_title(title, color="#0b0b0b", fontsize=12, pad=10)
    ax.set_xticks(np.arange(1, n_replicates + 1, 2))  # every 2nd replicate

    # --- light gridlines, only on the y-axis, kept unobtrusive ---
    ax.grid(True, axis="y", color="#e1e0d9", linewidth=0.8, zorder=0)

    # --- clean up the plot frame (no boxed-in look) ---
    for spine in ["top", "right"]:
        ax.spines[spine].set_visible(False)
    for spine in ["left", "bottom"]:
        ax.spines[spine].set_color("#c3c2b7")
    ax.tick_params(colors="#52514e")

    # --- legend: identifies each population by both color and shape ---
    ax.legend(frameon=False, loc="best")

    fig.tight_layout()
    fig.savefig(out_path, facecolor=fig.get_facecolor())
    plt.close(fig)
    print(f"Saved: {out_path}")


# ------------------------------------------------------------------------
# 4. MAKE THE TWO FIGURES
# ------------------------------------------------------------------------
plot_statistic_across_replicates(
    key_prefix="diversity",
    ylabel="Nucleotide diversity (π)",
    title="Nucleotide diversity across 20 replicates",
    out_path="diversity_replicates.png",
)

plot_statistic_across_replicates(
    key_prefix="TajimasD",
    ylabel="Tajima's D",
    title="Tajima's D across 20 replicates",
    out_path="tajimasd_replicates.png",
)
```
